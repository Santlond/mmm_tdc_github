```python
# Obtener nuevamente distribución de erroes bloqueantes experiencia digital tdc con las ingestiones más recientes
sql = """
SELECT nombre_experiencia,
       codigo_experiencia,
       codigo_error,
       descripcion_tecnica_error,
       descripcion_funcional_error,
       servicio_error,
       operacion_error,
       tipo_error,
       categoria_error,
       count(*) AS cantidad_registros
FROM resultados_vspc_canales.fco_logex_vd
WHERE YEAR BETWEEN 2024 AND 2026
  AND MONTH BETWEEN 1 AND 12
  AND DAY BETWEEN 1 AND 31
  AND cast(codigo_experiencia AS int) IN (10, -- preaprobado mc joven
 11, -- preaprobado mc ideal
 12, -- preaprobado multifranquicia
 21, -- mc negocios segm independientes
 180, -- preaprobado comportamental tdc unica virtual
 111, -- motor mc ideal
 110, -- motor mc joven
 112 -- motor multifranquicia
 )
  AND resultado_trx = 'ERROR'
  AND tipo_error NOT IN ('EXCEPCIONNEGOCIO',
                         'BUSINESS',
                         'NEGOCIO',
                         'EXCEPCIONES DE NEGOCIO',
                         'EXCEPCIONSERVICIO',
                         'BUSINESSEXCEPTIONS')
GROUP BY 1,
         2,
         3,
         4,
         5,
         6,
         7,
         8,
         9
ORDER BY count(*) DESC
"""
df_dist_errores_new = helper.obtener_dataframe(sql)
df_dist_errores_new
```

```python
# Convertir valores perdidos a cadena vacía en el dataframe df_dist_errores_new
df_dist_errores_new = df_dist_errores_new.fillna('')
```

```python
# Transformar todas las columnas del dataframe df_dist_errores_new a tipo string
df_dist_errores_new = df_dist_errores_new.astype(str)
df_dist_errores_new.info()
```

```python
df_dist_errores_old
```

```python
# Crear una columna adicional que se llame "llave_error" que sea la concatenación de las columnas nombre_experiencia, codigo_experiencia, codigo_error, descripcion_tecnica_error, descripcion_funcional_error, servicio_error, operacion_error, tipo_error y categoria_error
df_dist_errores_old['llave_error'] = df_dist_errores_old['nombre_experiencia'] + '_' + df_dist_errores_old['codigo_experiencia'] + '_' + df_dist_errores_old['codigo_error'] + '_' + df_dist_errores_old['descripcion_tecnica_error'] + '_' + df_dist_errores_old['descripcion_funcional_error'] + '_' + df_dist_errores_old['servicio_error'] + '_' + df_dist_errores_old['operacion_error'] + '_' + df_dist_errores_old['tipo_error'] + '_' + df_dist_errores_old['categoria_error']
df_dist_errores_new['llave_error'] = df_dist_errores_new['nombre_experiencia'] + '_' + df_dist_errores_new['codigo_experiencia'] + '_' + df_dist_errores_new['codigo_error'] + '_' + df_dist_errores_new['descripcion_tecnica_error'] + '_' + df_dist_errores_new['descripcion_funcional_error'] + '_' + df_dist_errores_new['servicio_error'] + '_' + df_dist_errores_new['operacion_error'] + '_' + df_dist_errores_new['tipo_error'] + '_' + df_dist_errores_new['categoria_error']
```

```python
# Adicionar la columna "error_bloqueante" del dataframe df_dist_errores_old al dataframe df_dist_errores_new
df_dist_errores_new = df_dist_errores_new.merge(df_dist_errores_old[['llave_error', 'error_bloqueante']], on='llave_error', how='left')
df_dist_errores_new.info()
```
